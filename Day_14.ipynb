```python
#anagram or not
s1=input("Enter first string:")
s2=input("Enter second string:")
if len(s1)==len(s2):
  a=sorted(s1.lower())
  b=sorted(s2.lower())
if a==b:
  print("Anagram")
else:
  print("Not Anagram")
```

```python
s = input("Enter a string: ")

upper = 0
lower = 0

for i in s:
    if i.isupper():
        upper += 1
    elif i.islower():
        lower += 1

print("Uppercase:", upper)
print("Lowercase:", lower)
```
